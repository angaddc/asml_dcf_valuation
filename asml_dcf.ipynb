## 2. Margins & Growth

Everything in this section is **calculated from the "1. Historical Financials" DataFrame above** —
no new hardcoded numbers are introduced.

- **Net sales YoY growth %** — this year's net sales divided by last year's, minus one. It measures
  how fast the top line is growing *period over period*, which is the natural way to look at growth:
  there's no FY2020 figure in our table, so FY2021 has nothing to compare against and is left blank
  (NaN).
- **Gross margin %**, **R&D as % of net sales**, **SG&A as % of net sales**, **Operating margin %**,
  and **Net margin %** are all *common-sized* figures: each line item is expressed as a percentage of
  that same year's net sales, rather than as a year-over-year change. Common-sizing is the right lens
  for margins because it strips out the effect of the business simply being bigger or smaller in a
  given year — it answers "of every euro of revenue, how much falls through to gross profit / R&D /
  operating income / net income?", which is comparable across years (or against other companies)
  regardless of absolute scale.

In short: **growth rates compare a line item to itself across time**, while **margins compare a line
item to revenue within the same year**. Both are needed — growth tells you the trajectory, margins
tell you the profitability and cost structure driving that trajectory.

```python
net_sales = historicals.loc["Net sales"]

margins_growth = pd.DataFrame(index=[
    "Net sales YoY growth %",
    "Gross margin %",
    "R&D as % of net sales",
    "SG&A as % of net sales",
    "Operating margin %",
    "Net margin %",
], columns=historicals.columns, dtype=float)

margins_growth.loc["Net sales YoY growth %"] = net_sales.pct_change() * 100
margins_growth.loc["Gross margin %"] = historicals.loc["Gross profit"] / net_sales * 100
margins_growth.loc["R&D as % of net sales"] = historicals.loc["R&D expenses"].abs() / net_sales * 100
margins_growth.loc["SG&A as % of net sales"] = historicals.loc["SG&A expenses"].abs() / net_sales * 100
margins_growth.loc["Operating margin %"] = historicals.loc["Operating income (EBIT)"] / net_sales * 100
margins_growth.loc["Net margin %"] = historicals.loc["Net income"] / net_sales * 100

# Formatted view for display; margins_growth itself keeps the underlying numeric values.
margins_growth_display = margins_growth.map(lambda v: f"{v:.1f}%" if pd.notna(v) else "–")
display(margins_growth_display)

# 5-year revenue CAGR, FY2021 -> FY2025 (4 periods of growth between 5 data points)
years_elapsed = 4
revenue_cagr = (net_sales["FY2025"] / net_sales["FY2021"]) ** (1 / years_elapsed) - 1
print(f"FY2021-FY2025 revenue CAGR: {revenue_cagr:.1%}")
```

**Important caveat: this is management's own framing, not independent analysis.** A CEO statement is

treated as a substitute for either.

- **Strengths / Weaknesses / Opportunities** are drawn from the CEO/President's statement in each

## 6. Forecast Financials & Unlevered FCF (FY2026–FY2030)

**This section is purely mechanical — every judgment call already happened in Section 5.** Nothing
here is hardcoded; every cell below is a formula that reads the percentage assumptions from Section 5
(and the FY2025 actual net sales from Section 1) and mechanically works through to unlevered free
cash flow (UFCF). If you want to change the forecast, go edit Section 5 — this section will update
automatically.

The build, step by step:
1. **Net sales** and **Gross profit** — compounding the Revenue growth % assumption forward from the
   FY2025 actual, then applying the Gross margin % assumption (same calculation Section 5's "Implied"
   table already did — reused here rather than recomputed).
2. **R&D expenses** and **SG&A expenses** — each year's Net sales × that year's R&D%/SG&A% assumption.
3. **EBIT** = Gross profit − R&D expenses − SG&A expenses — operating profit before interest and tax.
4. **Tax on EBIT** = EBIT × Effective tax rate % — the cash tax bill on operating profit, ignoring
   financing effects (this is the "unlevered" part: we tax EBIT directly, not profit after interest).
5. **NOPAT** (Net Operating Profit After Tax) = EBIT − Tax on EBIT — what operating profit would be if
   the company had no debt.
6. **D&A** is added back — it was already subtracted in arriving at EBIT/NOPAT, but it's a non-cash
   accounting charge, so it needs adding back to get to a *cash* flow figure.
7. **Capex** is subtracted — the real cash spent on property, plant, equipment and intangible assets,
   which never shows up in the income statement at all but very much shows up in the cash flow.
8. **Change in net working capital (ΔNWC)** is subtracted — cash tied up funding growth (more
   inventory, more receivables, etc. as the business gets bigger), separate from capex or D&A.
9. **Unlevered Free Cash Flow** = NOPAT + D&A − Capex − ΔNWC — the cash the business generates before
   any financing decisions (debt/equity), which is exactly the cash flow a DCF discounts at WACC to
   get Enterprise Value.

```python
forecast = pd.DataFrame(index=[
    "Net sales",
    "Gross profit",
    "R&D expenses",
    "SG&A expenses",
    "EBIT",
    "Tax on EBIT",
    "NOPAT",
    "D&A",
    "Capex",
    "Change in NWC",
    "Unlevered Free Cash Flow",
], columns=forecast_assumptions.columns, dtype=float)

# Net sales and Gross profit: reuse Section 5's "implied" figures directly (same numbers, not
# recomputed) rather than duplicating the compounding logic here.
forecast.loc["Net sales"] = implied.loc["Implied net sales"]
forecast.loc["Gross profit"] = implied.loc["Implied gross profit"]

# R&D and SG&A: this year's Net sales x that year's R&D%/SG&A% assumption from Section 5.
forecast.loc["R&D expenses"] = forecast.loc["Net sales"] * forecast_assumptions.loc["R&D % of net sales"] / 100
forecast.loc["SG&A expenses"] = forecast.loc["Net sales"] * forecast_assumptions.loc["SG&A % of net sales"] / 100

forecast.loc["EBIT"] = forecast.loc["Gross profit"] - forecast.loc["R&D expenses"] - forecast.loc["SG&A expenses"]

# Tax on EBIT and NOPAT: apply Section 5's effective tax rate assumption directly to EBIT (the
# "unlevered" step - tax is applied to operating profit, not profit after interest).
tax_rate = forecast_assumptions.loc["Effective tax rate %"] / 100
forecast.loc["Tax on EBIT"] = forecast.loc["EBIT"] * tax_rate
forecast.loc["NOPAT"] = forecast.loc["EBIT"] * (1 - tax_rate)

# D&A, Capex, Change in NWC: this year's Net sales x that year's %-of-net-sales assumption from
# Section 5. All three are stored as positive amounts here (D&A added back, Capex and NWC subtracted
# below), matching the formula as specified rather than the negative-expense convention used in
# Section 1's historical table.
forecast.loc["D&A"] = forecast.loc["Net sales"] * forecast_assumptions.loc["D&A % of net sales"] / 100
forecast.loc["Capex"] = forecast.loc["Net sales"] * forecast_assumptions.loc["Capex % of net sales"] / 100
forecast.loc["Change in NWC"] = forecast.loc["Net sales"] * forecast_assumptions.loc["Change in NWC % of net sales"] / 100

forecast.loc["Unlevered Free Cash Flow"] = (
    forecast.loc["NOPAT"] + forecast.loc["D&A"] - forecast.loc["Capex"] - forecast.loc["Change in NWC"]
)

# Sanity check: EBIT should split exactly into Tax on EBIT + NOPAT for every year.
assert (forecast.loc["Tax on EBIT"] + forecast.loc["NOPAT"] - forecast.loc["EBIT"]).abs().max() < 1e-6, \
    "EBIT does not split into Tax on EBIT + NOPAT"

forecast_display = forecast.map(lambda v: f"€{v:,.0f}m")
display(forecast_display)
```

## 7. WACC (Discount Rate)

The Weighted Average Cost of Capital (WACC) is the rate we'll use to discount the unlevered free
cash flows from Section 6 back to present value — it represents the blended return that both equity
holders and debt holders require to fund ASML, weighted by how much of the company is actually
funded by each.

**Every input below is hardcoded and individually sourced** (this section behaves like Sections 1 and
3 — sourced facts — rather than Section 5's judgment-call assumptions), with one explicit exception
flagged below. The four WACC formulas themselves (cost of equity, after-tax cost of debt, capital
weights, WACC) are all computed, not hardcoded.

**Key takeaway — capital structure is overwhelmingly equity-funded.** ASML's FY2025 long-term debt
(€2,709.0m, Section 3) is under 1% of its market capitalization (~€533bn). With debt weight that
small, WACC ends up very close to the cost of equity alone — the after-tax cost of debt barely moves
the blended number, no matter how it's estimated.

**The one estimated (not directly sourced) input is the credit spread used to build the pre-tax cost
of debt.** ASML's actual traded bond yields would be the correct input, but they sit behind paywalls
on every source checked; a ~0.70% credit spread over the risk-free rate (consistent with ASML's
Moody's A2 rating) is used instead as a reasoned estimate, clearly flagged in the code below. Given
how small debt's weight is in the blend, even a meaningfully wrong estimate here would barely move
the final WACC.

```python
# ==============================================================================================
#  WACC INPUTS - hardcoded, individually sourced (one estimate flagged explicitly below)
# ==============================================================================================

# Risk-free rate: German 10-year Bund yield, Sept 14 2026.
# Source: https://www.thetrading.tools/germany-10-year-yield ("Germany 10Y stands at 3.55%")
risk_free_rate = 3.55 / 100

# Beta: ASML's (levered) beta.
# Source: https://stockanalysis.com/stocks/asml/, Sept 14 2026 ("Beta: 1.36")
beta = 1.36

# Equity risk premium: Damodaran's Netherlands / mature-market total equity risk premium estimate.
# Source: https://pages.stern.nyu.edu/~adamodar/New_Home_Page/datafile/ctryprem.html
# (Netherlands row, 4.23%; page last updated January 2026)
equity_risk_premium = 4.23 / 100

# Market capitalization: built from two sourced primitives rather than hardcoding the converted
# EUR figure directly, so the conversion itself is a visible formula, not a re-typed number.
#   - Market cap in USD: https://stockanalysis.com/stocks/asml/, Sept 14 2026 ("Market Cap: 615.10B")
#   - EUR/USD exchange rate: https://tradingeconomics.com/euro-area/currency, Sept 14 2026 (~1.1537)
market_cap_usd_bn = 615.10
eur_usd_rate = 1.1537
market_cap_eur_m = market_cap_usd_bn * 1000 / eur_usd_rate  # -> ~EUR 533,100m

# Long-term debt: NOT re-hardcoded here - pulled directly from Section 3's already-sourced FY2025
# figure (cash_flow_bs), so this stays correct if Section 3 is ever revised.
long_term_debt_eur_m = cash_flow_bs.loc["Long-term debt (disclosed component only)", "FY2025"]

# Pre-tax cost of debt: built as Risk-free rate + an assumed credit spread, NOT independently
# sourced - ASML's actual traded bond yields were behind paywalls on every source checked.
# FLAGGED AS AN ESTIMATE: ~0.70% credit spread assumed, consistent with ASML's Moody's A2 rating.
credit_spread_estimate = 0.70 / 100  # ESTIMATE, not sourced - see markdown flag above
pre_tax_cost_of_debt = risk_free_rate + credit_spread_estimate

# Effective tax rate: NOT re-hardcoded - reused directly from Section 5's assumption (itself
# anchored to the Section 1 historical average), which is flat across the forecast so any column
# gives the same value.
tax_rate_for_wacc = forecast_assumptions.loc["Effective tax rate %", "FY2026"] / 100

# ==============================================================================================
#  WACC FORMULAS - all computed, nothing hardcoded below this line
# ==============================================================================================

cost_of_equity = risk_free_rate + beta * equity_risk_premium
after_tax_cost_of_debt = pre_tax_cost_of_debt * (1 - tax_rate_for_wacc)

total_capital = market_cap_eur_m + long_term_debt_eur_m
equity_weight = market_cap_eur_m / total_capital
debt_weight = long_term_debt_eur_m / total_capital

wacc = equity_weight * cost_of_equity + debt_weight * after_tax_cost_of_debt

wacc_inputs = pd.Series({
    "Risk-free rate": f"{risk_free_rate:.2%}",
    "Beta": f"{beta:.2f}",
    "Equity risk premium": f"{equity_risk_premium:.2%}",
    "Market capitalization (EUR m)": f"€{market_cap_eur_m:,.0f}m",
    "Long-term debt (EUR m, FY2025)": f"€{long_term_debt_eur_m:,.1f}m",
    "Pre-tax cost of debt (ESTIMATE)": f"{pre_tax_cost_of_debt:.2%}",
    "Effective tax rate": f"{tax_rate_for_wacc:.1%}",
})
print("WACC inputs:")
display(wacc_inputs)

wacc_outputs = pd.Series({
    "Cost of equity (CAPM)": f"{cost_of_equity:.2%}",
    "After-tax cost of debt": f"{after_tax_cost_of_debt:.2%}",
    "Equity weight": f"{equity_weight:.2%}",
    "Debt weight": f"{debt_weight:.2%}",
    "WACC": f"{wacc:.2%}",
})
print("\nWACC build:")
display(wacc_outputs)

print(f"\nWACC = {wacc:.2%} (vs. cost of equity alone = {cost_of_equity:.2%}) "
      f"- nearly identical, since debt weight is only {debt_weight:.2%} of total capital.")
```

## 8. Terminal Value (Gordon Growth vs Exit Multiple)

The 5-year explicit forecast (Sections 5–6) can't run forever — at some point we need a **terminal
value** that stands in for all the cash flows ASML generates beyond FY2030. This section builds that
terminal value two different ways and cross-checks them against each other, per the project's
approach of always building both methods rather than relying on one.

- **Gordon Growth (perpetuity growth) method** assumes FY2030's unlevered FCF grows forever at a
  constant, modest rate, discounted at WACC. It's a *cash-flow-based* method.
- **Exit Multiple method** assumes ASML gets sold, at the end of FY2030, for a multiple of its EBITDA
  that a buyer would pay today for a similar business. It's a *market-based* method, anchored to how
  the market actually prices comparable(-to-ASML-today) companies rather than a growth assumption.

**Expect these two methods to disagree, and expect the gap to be large here.** A textbook "mature,
stable" company would see the two converge, because a market multiple on a mature business is itself
just a compressed way of pricing in a modest perpetual growth rate — similar to what Gordon Growth
assumes directly. ASML's 10-year median EV/EBITDA multiple (used deliberately here, instead of
today's elevated ~44x, specifically to avoid extrapolating a possibly-temporary AI-driven re-rating)
still implies a *much* higher perpetual growth rate than the ECB's 2% medium-term inflation target
used for Gordon Growth — because the market has historically been pricing ASML as a structural
growth business (EUV/High-NA monopoly, secular semiconductor capex growth), not a mature/stable one.
**A large gap here is the expected, correct result of that difference in method — not a bug to
reconcile away.** The code cell below explicitly back-solves the perpetual growth rate the exit
multiple is implying, so that gap is visible as a number, not just a vibe.

```python
# ==============================================================================================
#  TERMINAL VALUE INPUTS - hardcoded, individually sourced
# ==============================================================================================

# Terminal growth rate: ECB's medium-term inflation target.
# Source: https://www.ecb.europa.eu/mopo/strategy/pricestab/html/index.en.html
# ("the ECB's Governing Council ... considers that price stability is best maintained by aiming
# for 2% inflation over the medium term")
terminal_growth_rate = 2.0 / 100

# Exit EV/EBITDA multiple: ASML's 10-year historical median (deliberately used instead of the
# current ~44x multiple, to avoid extrapolating today's possibly-elevated, AI-driven re-rating
# into a permanent assumption).
# Source: https://www.gurufocus.com/term/enterprise-value-to-ebitda/ASML, Sept 15 2026
# (page states ASML's current EV/EBITDA of 44.00 is "51% Above Median", with 10-year median 29.06 -
# 44.00 / 1.51 = 29.14, consistent with the stated 29.06 median within rounding)
exit_ev_ebitda_multiple = 29.06

# ==============================================================================================
#  TERMINAL VALUE FORMULAS - all computed from Section 6 (forecast) and Section 7 (wacc)
# ==============================================================================================

fy2030_ufcf = forecast.loc["Unlevered Free Cash Flow", "FY2030"]
fy2030_ebitda = forecast.loc["EBIT", "FY2030"] + forecast.loc["D&A", "FY2030"]

# Gordon Growth (perpetuity growth) terminal value, as of end of FY2030.
gordon_growth_tv = fy2030_ufcf * (1 + terminal_growth_rate) / (wacc - terminal_growth_rate)

# Exit Multiple terminal value, as of end of FY2030.
exit_multiple_tv = fy2030_ebitda * exit_ev_ebitda_multiple

# Cross-check: back-solve the Gordon Growth formula for the perpetual growth rate that would be
# needed to justify the Exit Multiple terminal value, given the same FY2030 UFCF and WACC.
#   TV = FCF*(1+g) / (WACC-g)  =>  g = (TV*WACC - FCF) / (TV + FCF)
implied_growth_from_exit_multiple = (
    (exit_multiple_tv * wacc - fy2030_ufcf) / (exit_multiple_tv + fy2030_ufcf)
)

terminal_value_summary = pd.Series({
    "FY2030 Unlevered FCF": f"€{fy2030_ufcf:,.0f}m",
    "FY2030 EBITDA (EBIT + D&A)": f"€{fy2030_ebitda:,.0f}m",
    "WACC (Section 7)": f"{wacc:.2%}",
    "Terminal growth rate (Gordon Growth)": f"{terminal_growth_rate:.2%}",
    "Gordon Growth terminal value": f"€{gordon_growth_tv:,.0f}m",
    "Exit EV/EBITDA multiple": f"{exit_ev_ebitda_multiple:.2f}x",
    "Exit Multiple terminal value": f"€{exit_multiple_tv:,.0f}m",
    "Implied perpetual growth rate from Exit Multiple": f"{implied_growth_from_exit_multiple:.2%}",
})
display(terminal_value_summary)

gap_x = exit_multiple_tv / gordon_growth_tv
print(f"\nExit Multiple terminal value is {gap_x:.1f}x the Gordon Growth terminal value.")
print(f"The exit multiple is implicitly pricing in ~{implied_growth_from_exit_multiple:.1%} perpetual "
      f"growth, vs. the {terminal_growth_rate:.1%} used directly in the Gordon Growth method.")
```

## 9. Valuation: EV to Equity Value to Per-Share

This is the final bridge: discount the explicit FY2026–FY2030 unlevered FCFs (Section 6) and both
Section 8 terminal values back to present value at WACC (Section 7), sum to Enterprise Value, then
walk EV down to an equity value and a per-share figure — **twice**, once per terminal value method.

**Why a range, not one number.** Gordon Growth and Exit Multiple aren't two attempts at the same
answer where one is "right" — they encode two different beliefs about what ASML becomes after 2030:
- The **Gordon Growth price is what ASML is worth if it settles into a mature, roughly
  GDP/inflation-rate-growing business** after 2030 — the 2% ECB inflation-anchored terminal growth
  rate from Section 8 treats FY2030 as basically "steady state."
- The **Exit Multiple price is what ASML is worth if the market keeps pricing it as a durable
  secular grower** — Section 8 showed this method's 10-year-median multiple is implicitly pricing in
  ~6.8% perpetual growth, well above GDP/inflation, consistent with ASML's structural EUV/High-NA
  position rather than a maturing business.

Both are defensible; which one is closer to reality depends on a view of ASML's competitive position
in the 2030s that this notebook hasn't tried to independently adjudicate (that's exactly the kind of
qualitative judgment Section 4 flagged as "one input among several," not something a spreadsheet
settles). **The honest output of this model is therefore the range between the two, not a single
blended midpoint** — collapsing it into one number would hide that those two beliefs disagree by a
lot, which is real information, not noise to average away.

**Net debt sign check:** ASML holds far more cash (€13,321.9m, Section 3, FY2025) than long-term debt
(€2,709.0m, Section 7), so it's in a **net cash position** — net debt is *negative*. The bridge below
uses the standard formula Equity Value = EV − Net Debt with Net Debt = Debt − Cash; because Net Debt
comes out negative here, subtracting it *adds* the net cash to EV automatically — the formula doesn't
need special-casing, but the code below prints the net debt figure explicitly so that sign is visible
rather than just implicit in the arithmetic.

```python
# ==============================================================================================
#  DISCOUNT EXPLICIT-PERIOD FCF AND BOTH TERMINAL VALUES TO PRESENT VALUE
# ==============================================================================================

forecast_years = forecast.columns.tolist()  # FY2026..FY2030
discount_periods = pd.Series(range(1, len(forecast_years) + 1), index=forecast_years)  # year 1..5
discount_factors = 1 / (1 + wacc) ** discount_periods

pv_fcf = forecast.loc["Unlevered Free Cash Flow"] * discount_factors

pv_build = pd.DataFrame({
    "Unlevered FCF": forecast.loc["Unlevered Free Cash Flow"],
    "Discount period": discount_periods,
    "Discount factor": discount_factors,
    "PV of FCF": pv_fcf,
})
pv_build_display = pv_build.copy()
pv_build_display["Unlevered FCF"] = pv_build_display["Unlevered FCF"].map(lambda v: f"€{v:,.0f}m")
pv_build_display["Discount factor"] = pv_build_display["Discount factor"].map(lambda v: f"{v:.4f}")
pv_build_display["PV of FCF"] = pv_build_display["PV of FCF"].map(lambda v: f"€{v:,.0f}m")
print("Present value of each explicit-period FCF (Section 6 FCF x discount factor at WACC):")
display(pv_build_display.T)

sum_pv_fcf = pv_fcf.sum()

# Both terminal values (Section 8) are "as of end of FY2030" - same discount factor as the FY2030
# FCF (year 5), per the brief.
year5_discount_factor = discount_factors["FY2030"]
pv_gordon_growth_tv = gordon_growth_tv * year5_discount_factor
pv_exit_multiple_tv = exit_multiple_tv * year5_discount_factor

ev_gordon_growth = sum_pv_fcf + pv_gordon_growth_tv
ev_exit_multiple = sum_pv_fcf + pv_exit_multiple_tv

print(f"\nSum of PV of explicit-period FCFs: EUR {sum_pv_fcf:,.0f}m")
print(f"PV of Gordon Growth terminal value: EUR {pv_gordon_growth_tv:,.0f}m")
print(f"PV of Exit Multiple terminal value: EUR {pv_exit_multiple_tv:,.0f}m")
print(f"\nEnterprise Value (Gordon Growth method): EUR {ev_gordon_growth:,.0f}m")
print(f"Enterprise Value (Exit Multiple method):  EUR {ev_exit_multiple:,.0f}m")

# ==============================================================================================
#  EV -> EQUITY VALUE -> PER-SHARE, FOR BOTH METHODS
# ==============================================================================================

# Long-term debt: Section 7 (itself sourced from Section 3, FY2025).
# Cash & cash equivalents: Section 3, FY2025 (includes short-term investments).
debt_eur_m = long_term_debt_eur_m
cash_eur_m = cash_flow_bs.loc["Cash & cash equivalents (incl. short-term investments)", "FY2025"]

# Net debt = Debt - Cash. ASML holds far more cash than debt, so this comes out NEGATIVE (net cash).
# Equity Value = EV - Net Debt handles the sign correctly on its own: subtracting a negative number
# adds the net cash back to EV, with no special-casing required.
net_debt_eur_m = debt_eur_m - cash_eur_m
print(f"\nLong-term debt (Section 7): EUR {debt_eur_m:,.1f}m")
print(f"Cash & equivalents (Section 3, FY2025): EUR {cash_eur_m:,.1f}m")
print(f"Net debt = Debt - Cash = EUR {net_debt_eur_m:,.1f}m "
      f"({'NET CASH position (negative net debt)' if net_debt_eur_m < 0 else 'net debt position'})")

diluted_shares_m = cash_flow_bs.loc["Weighted avg shares - diluted (millions)", "FY2025"]

equity_value_gordon = ev_gordon_growth - net_debt_eur_m
equity_value_exit = ev_exit_multiple - net_debt_eur_m

value_per_share_gordon = equity_value_gordon / diluted_shares_m
value_per_share_exit = equity_value_exit / diluted_shares_m

valuation_bridge = pd.DataFrame({
    "Gordon Growth method": {
        "Enterprise Value": f"€{ev_gordon_growth:,.0f}m",
        "less: Net debt": f"€{net_debt_eur_m:,.1f}m",
        "= Equity Value": f"€{equity_value_gordon:,.0f}m",
        "/ Diluted shares (m)": f"{diluted_shares_m:.1f}m",
        "= Value per share": f"€{value_per_share_gordon:,.2f}",
    },
    "Exit Multiple method": {
        "Enterprise Value": f"€{ev_exit_multiple:,.0f}m",
        "less: Net debt": f"€{net_debt_eur_m:,.1f}m",
        "= Equity Value": f"€{equity_value_exit:,.0f}m",
        "/ Diluted shares (m)": f"{diluted_shares_m:.1f}m",
        "= Value per share": f"€{value_per_share_exit:,.2f}",
    },
})
print()
display(valuation_bridge)
```

## 10. Sensitivity Analysis

Section 9 produced two single-point implied values per share, each resting on a specific WACC and
either a specific terminal growth rate or a specific exit multiple. This section stress-tests both
of those single points by varying the two inputs each method is most sensitive to, holding
everything else fixed at Section 6's forecast and Section 9's net debt / share count — the same
full EV → Equity Value → per-share formula as Section 9, just re-run across a grid of WACC and
terminal-value inputs instead of at one base case.

**Color coding:** green cells imply a value *above* the current market price (€1,406.80, Section 9);
red cells imply a value *below* it. **The bordered cell in each grid is the base case** — the same
WACC (~9%, actually 9.27%) and terminal assumption (2.0% growth / 29.06x multiple) used in Sections
8–9 — so you can see at a glance how much the single-point answer moves if WACC or the terminal
assumption is even modestly different from what was assumed there.

### Grid 1 — Gordon Growth method

Implied value per share as WACC (rows) and the Gordon Growth terminal growth rate (columns) vary.
This is the more conservative of the two methods (Section 8/9) — **every cell in this grid is red**:
even the most favorable combination tested (WACC 7%, terminal growth 3.0%) only implies €1,047 per
share, still 25.6% below the current €1,406.80 market price. The Gordon Growth method can't get
close to the current market price anywhere in this parameter range, which is itself informative
(see the summary below).

```python
def implied_value_per_share_gordon(wacc_input, terminal_growth_input):
    """Same EV -> Equity Value -> per-share formula as Section 9's Gordon Growth method, with
    WACC and terminal growth rate as free parameters. Reuses Section 6's forecast FCF and
    Section 9's net debt / diluted share count - nothing else changes."""
    periods = pd.Series(range(1, len(forecast.columns) + 1), index=forecast.columns)
    discount_factors_local = 1 / (1 + wacc_input) ** periods
    pv_fcf_local = (forecast.loc["Unlevered Free Cash Flow"] * discount_factors_local).sum()
    tv_local = (
        forecast.loc["Unlevered Free Cash Flow", "FY2030"]
        * (1 + terminal_growth_input)
        / (wacc_input - terminal_growth_input)
    )
    pv_tv_local = tv_local * discount_factors_local["FY2030"]
    ev_local = pv_fcf_local + pv_tv_local
    equity_value_local = ev_local - net_debt_eur_m
    return equity_value_local / diluted_shares_m


wacc_grid_values = [0.07, 0.08, 0.09, 0.10, 0.11, 0.12]
terminal_growth_grid_values = [0.005, 0.010, 0.015, 0.020, 0.025, 0.030]

wacc_row_labels = [f"{w:.0%}" for w in wacc_grid_values]
terminal_growth_col_labels = [f"{g:.1%}" for g in terminal_growth_grid_values]

gordon_grid = pd.DataFrame(index=wacc_row_labels, columns=terminal_growth_col_labels, dtype=float)
for w, w_label in zip(wacc_grid_values, wacc_row_labels):
    for g, g_label in zip(terminal_growth_grid_values, terminal_growth_col_labels):
        gordon_grid.loc[w_label, g_label] = implied_value_per_share_gordon(w, g)

print("Gordon Growth grid - raw values (EUR per share):")
display(gordon_grid.map(lambda v: f"{v:,.0f}"))

# Base case: WACC 9.27% (Section 7) rounds to the "9%" row; terminal growth 2.0% (Section 8) is the
# "2.0%" column exactly.
gordon_base_row = "9%"
gordon_base_col = "2.0%"


def highlight_vs_market(v):
    color = "#c6efce" if v > current_share_price_eur else "#ffc7ce"  # pastel green / red
    return f"background-color: {color}"


gordon_styled = (
    gordon_grid.style
    .map(highlight_vs_market)
    .format("€{:,.0f}")
    .set_properties(
        subset=pd.IndexSlice[[gordon_base_row], [gordon_base_col]],
        **{"border": "3px solid black", "font-weight": "bold"},
    )
    .set_caption("Gordon Growth: implied EUR value per share (rows = WACC, columns = terminal growth rate)")
)
display(gordon_styled)
```

### Grid 2 — Exit Multiple method

Implied value per share as WACC (rows) and the exit EV/EBITDA multiple (columns) vary. This grid is
mostly green: the base-case cell (WACC ~9%, 29.06x) and everything to its right (higher multiples)
or above it (lower WACC) clears the current market price. The red pocket sits at low multiples
combined with higher WACC — the 20x column is red at every WACC tested, and 25x turns red once WACC
reaches 8% or higher.

```python
def implied_value_per_share_exit(wacc_input, exit_multiple_input):
    """Same EV -> Equity Value -> per-share formula as Section 9's Exit Multiple method, with WACC
    and the exit EV/EBITDA multiple as free parameters. Reuses Section 6's forecast FCF/EBITDA and
    Section 9's net debt / diluted share count - nothing else changes."""
    periods = pd.Series(range(1, len(forecast.columns) + 1), index=forecast.columns)
    discount_factors_local = 1 / (1 + wacc_input) ** periods
    pv_fcf_local = (forecast.loc["Unlevered Free Cash Flow"] * discount_factors_local).sum()
    tv_local = fy2030_ebitda * exit_multiple_input
    pv_tv_local = tv_local * discount_factors_local["FY2030"]
    ev_local = pv_fcf_local + pv_tv_local
    equity_value_local = ev_local - net_debt_eur_m
    return equity_value_local / diluted_shares_m


exit_multiple_grid_values = [20, 25, 29.06, 35, 40]
exit_multiple_col_labels = [f"{m:g}x" for m in exit_multiple_grid_values]

exit_grid = pd.DataFrame(index=wacc_row_labels, columns=exit_multiple_col_labels, dtype=float)
for w, w_label in zip(wacc_grid_values, wacc_row_labels):
    for m, m_label in zip(exit_multiple_grid_values, exit_multiple_col_labels):
        exit_grid.loc[w_label, m_label] = implied_value_per_share_exit(w, m)

print("Exit Multiple grid - raw values (EUR per share):")
display(exit_grid.map(lambda v: f"{v:,.0f}"))

# Base case: WACC 9.27% (Section 7) rounds to the "9%" row; exit multiple 29.06x (Section 8) is the
# "29.06x" column exactly.
exit_base_row = "9%"
exit_base_col = "29.06x"

exit_styled = (
    exit_grid.style
    .map(highlight_vs_market)
    .format("€{:,.0f}")
    .set_properties(
        subset=pd.IndexSlice[[exit_base_row], [exit_base_col]],
        **{"border": "3px solid black", "font-weight": "bold"},
    )
    .set_caption("Exit Multiple: implied EUR value per share (rows = WACC, columns = exit EV/EBITDA multiple)")
)
display(exit_styled)
```

### Where the current market price sits in each grid

- **Gordon Growth grid: every single cell is below the current market price.** Even the most
  favorable combination tested — WACC 7% (a point or more below Section 7's 9.27% base case) and
  3.0% terminal growth (well above the 2% ECB-inflation-anchored rate actually used) — only reaches
  €1,047, still 25.6% below €1,406.80. **The market price is not explainable by the Gordon Growth
  method anywhere in this grid**, not just at the base case — reinforcing, more starkly than Section
  9's single point (−57.9%) already suggested, that the market is not pricing ASML as a "matures to
  a stable, modest-growth business" story.
- **Exit Multiple grid: the current price sits comfortably inside the green region, not at its
  edge.** At the base-case WACC (~9%) and ASML's own 10-year median multiple (29.06x), the implied
  value (€1,477 in this grid; €1,459.47 exactly in Section 9) already clears the market price. The
  grid only turns red at low multiples (20x is red at every WACC tested; 25x turns red once WACC
  reaches 8% or higher) — both well below the *current* traded multiple of ~39–44x noted in Section
  8, which would push even further into green.
- **Taken together**, the two grids sharpen Section 9's core point rather than soften it: it isn't
  that the Gordon Growth base case is merely somewhat conservative — across this entire tested range
  it can't reach the current price at all. The market's valuation of ASML is really only explicable
  through something closer to the Exit Multiple method's implicit growth story.
